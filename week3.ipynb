```python
import pandas as pd
data = {
    "name" : ["Ali", "Khan" , "Hussain" ],
    "Age": [21,24,21],
    "GPA" : [3.2,3.3,4.5],
    "major":  ["CS","AI","Stats"],
    "Gender": ["M","F","M"]
}
df =pd.DataFrame(data)
print(df)
```

```python
print(df.isna().sum())
print(df.isna())
print(df.sum())

```

```python
clean = df.dropna()
print(clean)
```

```python
df["GPA"] = df["GPA"].fillna(df["GPA"].mean())
print(df)
```

```python
df["major"]=df["major"].fillna("Unkown")
```
